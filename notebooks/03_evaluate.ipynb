# 03 - Evaluation

Written and run only after the final configuration is decided and written down in the work log. This is the only moment the test set is touched - every early peek at it turns it into something you tuned toward, and then it is no longer a clean measurement.

What happens here: run the model on the test set, compute the scores, draw the plots for the report, and read 15-20 real mistakes by hand to understand why they happened.
